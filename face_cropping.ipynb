```python
import cv2
import os

def crop_faces_recursive(root_folder, output_root, min_size=(80, 80)):
    cascade = cv2.CascadeClassifier(
        cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    )

    # Walk through all subfolders
    for split in os.listdir(root_folder):     # train, val, test
        split_path = os.path.join(root_folder, split)

        if not os.path.isdir(split_path):
            continue

        for cls in os.listdir(split_path):    # AN, DI, etc.
            class_path = os.path.join(split_path, cls)

            if not os.path.isdir(class_path):
                continue

            # Create matching output directory
            out_class_path = os.path.join(output_root, split, cls)
            os.makedirs(out_class_path, exist_ok=True)

            # Process all images in this class folder
            for file_name in os.listdir(class_path):
                if not file_name.lower().endswith(('.tiff')):
                    continue

                img_path = os.path.join(class_path, file_name)
                img = cv2.imread(img_path)
                if img is None:
                    print(f"Could not read {img_path}, skipping...")
                    continue

                gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

                faces = cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=min_size)

                if len(faces) == 0:
                    print(f"No face found in {img_path}")
                    continue

                # Largest face
                x, y, w, h = sorted(faces, key=lambda f: f[2] * f[3], reverse=True)[0]
                face_crop = img[y:y+h, x:x+w]

                save_path = os.path.join(out_class_path, file_name)
                cv2.imwrite(save_path, face_crop)
                print(f"Saved -> {save_path}")

    print("✅ Completed face cropping!")
```

```python
# ✅ Example usage
input_folder = "/Users/ricardosantiago/Documents/GitHub/neurodocs/CMANAS/datasets/loso_cropping/KA/"
output_folder = "/Users/ricardosantiago/Documents/GitHub/neurodocs/CMANAS/datasets/loso_cropping/KA/"

crop_faces_recursive(input_folder, output_folder)
```
